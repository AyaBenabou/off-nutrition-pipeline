```python
import pydata_google_auth
from google.cloud import bigquery

credentials = pydata_google_auth.get_user_credentials(
    ["https://www.googleapis.com/auth/bigquery"],
)
client = bigquery.Client(project="off-nutrition-pipeline", credentials=credentials)

sql = """
SELECT
  p.pays_nom,
  COUNT(*) AS nb_produits,
  COUNTIF(f.nutriscore_lettre IS NOT NULL) AS nb_notes,
  ROUND(100 * COUNTIF(f.nutriscore_lettre IN ('a','b'))
        / NULLIF(COUNTIF(f.nutriscore_lettre IS NOT NULL), 0), 1) AS pct_a_b,
  ROUND(100 * COUNTIF(f.nutriscore_lettre IN ('d','e'))
        / NULLIF(COUNTIF(f.nutriscore_lettre IS NOT NULL), 0), 1) AS pct_d_e
FROM `off-nutrition-pipeline.gold.fait_produit_pays` f
JOIN `off-nutrition-pipeline.gold.dim_pays` p USING (pays_id)
WHERE p.pays_retenu
GROUP BY p.pays_nom
ORDER BY pct_a_b DESC
"""
print(f"{'Pays':<24}{'Produits':>10}{'Notés':>10}{'% A-B':>8}{'% D-E':>8}")
for l in client.query(sql).result():
    print(f"{l['pays_nom']:<24}{l['nb_produits']:>10}{l['nb_notes']:>10}{l['pct_a_b']:>8}{l['pct_d_e']:>8}")
```
